### Objective: Loading → inspecting → cleaning → validating → merging → saving the master dataset

```python
import pandas as pd
import os
data_path = "../data"
files = os.listdir(data_path)
for file in files:
    file_path = os.path.join(data_path, file)
    print("\n" + "=" * 80)
    print(file)
    print("=" * 80)
    # Read without assuming a header
    raw = pd.read_excel(file_path, header=None)
    # Show first 10 rows
    display(raw.head(10))
```

```python
# Correct header rows identified from the raw Excel sheets

header_rows = {
    "DS1_plot_registry.xlsx": 1,
    "DS2_TSH_Records.xlsx": 1,
    "DS6_Precipitaion.xlsx": 1,
    "DS11_Nitrogen_Thesis.xlsx": 3,
    "DS12_Phosphorus_Thesis.xlsx": 3,
    "DS13_Potassium_Thesis.xlsx": 3,
    "DS14_GNDVI_Derived_Thesis.xlsx": 3,
    "DS17_Chlorophyll_Thesis.xlsx": 6
}

datasets = {}

for file, header_row in header_rows.items():
    file_path = os.path.join(data_path, file)
    datasets[file] = pd.read_excel(file_path, header=header_row)

print("All datasets loaded with correct headers.")
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)
    print(df.columns.tolist())
```

```python
for name, df in datasets.items():

    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)

    print("Shape:", df.shape)

    # Missing values
    missing = df.isna().sum()
    missing = missing[missing > 0]

    print("\nMissing values:")
    if len(missing) == 0:
        print("No missing values")
    else:
        print(missing)

    # Duplicate rows
    print("\nDuplicate rows:", df.duplicated().sum())

    # Plot ID information
    plot_col = "Plot_ID" if "Plot_ID" in df.columns else "Plot ID"

    print("Unique Plot IDs:", df[plot_col].nunique())
    print("Total rows:", len(df))
```

```python
plot_id_sets = {}

for name, df in datasets.items():

    plot_col = "Plot_ID" if "Plot_ID" in df.columns else "Plot ID"

    plot_id_sets[name] = set(df[plot_col].dropna().astype(str).str.strip())

print("Plot ID consistency check\n")

for name, ids in plot_id_sets.items():
    print(f"{name}: {len(ids)} unique Plot IDs")

print("\n" + "=" * 80)
print("COMMON PLOT IDs ACROSS ALL DATASETS")
print("=" * 80)

common_ids = set.intersection(*plot_id_sets.values())

print("Number of common Plot IDs:", len(common_ids))
print("First 20 common IDs:", sorted(common_ids)[:20])
```

```python
# Compare Plot IDs between every dataset and the common set

for name, ids in plot_id_sets.items():
    missing_from_common = ids - common_ids

    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)

    print("Plot IDs not present in ALL datasets:", len(missing_from_common))
    print(sorted(missing_from_common))
```

```python
all_plot_ids = sorted(set.union(*plot_id_sets.values()))

coverage = pd.DataFrame({
    "Plot_ID": all_plot_ids
})

for name, ids in plot_id_sets.items():
    short_name = name.replace(".xlsx", "")
    coverage[short_name] = coverage["Plot_ID"].isin(ids)

coverage["Dataset_Count"] = coverage.iloc[:, 1:].sum(axis=1)

display(coverage)
```

```python
# Standardize column names and remove metadata rows

for name, df in datasets.items():

    # Remove leading/trailing spaces from column names
    df.columns = df.columns.str.strip()

    # Standardize Plot ID column name
    if "Plot ID" in df.columns:
        df.rename(columns={"Plot ID": "Plot_ID"}, inplace=True)

    # Keep only actual plot records
    datasets[name] = df[
        datasets[name]["Plot_ID"].astype(str).str.match(r"^WC-\d{3}$", na=False)
    ].copy()

print("Metadata rows removed and Plot_ID standardized.")
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows | {df['Plot_ID'].nunique()} unique Plot IDs")
```

```python
plot_id_sets = {
    name: set(df["Plot_ID"])
    for name, df in datasets.items()
}

common_ids = set.intersection(*plot_id_sets.values())

print("Common Plot IDs across all 8 datasets:", len(common_ids))
print("Range:", min(common_ids), "to", max(common_ids))
print("First 10:", sorted(common_ids)[:10])
print("Last 10:", sorted(common_ids)[-10:])
```

### Inspect missing values and data types

```python
for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)
    print("Shape:", df.shape)
    # Missing values
    missing = df.isna().sum()
    missing = missing[missing > 0]
    print("\nMissing values:")
    if missing.empty:
        print("None")
    else:
        print(missing)
    # Data types
    print("\nData types:")
    print(df.dtypes)
```

### Check the important numerical variables

```python
for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)
    numeric_cols = df.select_dtypes(include="number").columns
    print("Numerical columns:")
    print(list(numeric_cols))
    if len(numeric_cols) > 0:
        display(df[numeric_cols].describe().T)
```

### A few important findings

Most datasets have no missing values in their meaningful measurement columns.
Notes in DS1 and DS17 is entirely empty, so we can ignore it.
DS17 has 100 observations, while the other datasets have 120. This is expected from our earlier ID check.
Some variables are essentially alternate representations of the same measurement. For example, N content pct DW and Kjeldahl N mg kg are directly related, and similarly for P and K. We should not blindly use every numerical column as a separate feature later.
DS14 EVI has a maximum of 6.5818, considerably above its 75th percentile of 1.18045. That is something we should investigate rather than automatically deleting.
DS2 TCH_t_per_ha and TSH_t_per_ha also have relatively high maximums compared with their medians, so we'll inspect their distributions before deciding whether they are genuine values or outliers.

### Check categorical variables

```python
for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)
    categorical_cols = df.select_dtypes(include="object").columns
    for col in categorical_cols:
        print(f"\n{col}:")
        print("Unique values:", df[col].nunique())
        print(df[col].value_counts(dropna=False).head(20))
```

### Check date columns

```python
for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)
    date_candidates = [
        col for col in df.columns
        if "date" in col.lower()
    ]
    print("Date columns:", date_candidates)
    for col in date_candidates:
        print(f"\n{col}")
        print(df[col].head())
```

### Check duplicate Plot IDs

```python
for name, df in datasets.items():

    duplicate_ids = df[df["Plot_ID"].duplicated(keep=False)]

    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)

    if duplicate_ids.empty:
        print("No duplicate Plot_IDs.")
    else:
        print("Duplicate Plot_IDs found:")
        display(duplicate_ids[["Plot_ID"]])
```

### Check potential outliers

```python
for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)
    numeric_cols = df.select_dtypes(include="number").columns
    for col in numeric_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        outliers = df[(df[col] < lower) | (df[col] > upper)]
        if len(outliers) > 0:
            print(
                f"{col}: {len(outliers)} potential outlier(s) "
                f"| Lower = {lower:.3f} | Upper = {upper:.3f}"
            )
```

### Convert date columns

```python
for name, df in datasets.items():
    date_cols = [
        col for col in df.columns
        if "date" in col.lower()
    ]
    for col in date_cols:
        datasets[name][col] = pd.to_datetime(
            datasets[name][col],
            errors="coerce"
        )
print("Date columns converted successfully.")

for name, df in datasets.items():

    date_cols = [
        col for col in df.columns
        if "date" in col.lower()
    ]

    if date_cols:
        print("\n" + name)
        print(df[date_cols].dtypes)
```

### Verify the analytical dataset size

```python
plot_id_sets = {
    name: set(df["Plot_ID"])
    for name, df in datasets.items()
}

common_ids = set.intersection(*plot_id_sets.values())

print("Number of common Plot IDs:", len(common_ids))
print("First 10:", sorted(common_ids)[:10])
print("Last 10:", sorted(common_ids)[-10:])
```

### The potential outliers are not automatically errors. We'll keep them for now.
Date columns were successfully converted to datetime64[ns].
Most importantly, we confirmed 100 common plots (WC-001–WC-100) across all 8 datasets.
The DS17 limitation is therefore clear: our integrated 8-dataset analysis will use these 100 plots.

### Merge DS1 + DS2 + DS6

```python
master = datasets["DS1_plot_registry.xlsx"].copy()
print("Starting master shape:", master.shape)
```
